```python
import pandas as pd
df = pd.read_csv("Dataset.csv")
df.head()
```

```python
#check missing values in each column
df.isnull().sum()
```

```python
# Total missing values in the dataset
print("Total missing values:",
      df.isnull().sum().sum())
```

```python
for col in ["director","country","rating","duration","date_added"]:
    if col in df.columns:
        count= (
  df[col].astype("string")
      .str.strip()
      .str.lower()
      .eq("not given")
      .sum()
        )
print(col,":",count)
```

```python
# Convert "Not Given" into missing values
cols = ["director", "country", "rating", "duration", "date_added"]

for col in cols:
    if col in df.columns:
        df[col] = df[col].replace(
            r"(?i)^\s*not given\s*$",
            pd.NA,
            regex=True
        )

# Fill descriptive columns with Unknown
for col in ["director", "country", "rating", "duration"]:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")

# Verify remaining missing values
print(df.isnull().sum())
```

```python
# Check duplicate records
print("Total duplicate rows:", df.duplicated().sum())
```

```python
# Remove exact duplicate rows
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
# Clean extra spaces in text columns
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].str.strip()
    df[col] = df[col].str.replace(r"\s+", " ", regex=True)

print("Formatting cleaned successfully!")
```

```python
# Check existing Type values
print(df["type"].value_counts(dropna=False))
```

```python
# Standardize Type column
df["type"] = df["type"].str.strip().str.lower()

df["type"] = df["type"].replace({
    "movie": "Movie",
    "tv show": "TV Show"
})

print(df["type"].value_counts(dropna=False))
```

```python
# Standardize Rating column
df["rating"] = df["rating"].apply(
    lambda x: x.upper() if isinstance(x, str) and x != "Unknown" else x
)

print(df["rating"].value_counts(dropna=False).head(10))
```

```python
# Standardize Country formatting
df["country"] = df["country"].apply(
    lambda x: ", ".join(part.strip() for part in x.split(","))
    if isinstance(x, str) and x != "Unknown"
    else x
)

print(df["country"].head(10))
```

```python
# Step 5: Export the cleaned dataset

df.to_csv("Dataset_Cleaned.csv", index=False)

print("Dataset exported successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset_Cleaned.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
content_counts = df["type"].value_counts()

print("Content Distribution:")
print(content_counts)

print("\nTotal Movies:", content_counts.get("Movie", 0))
print("Total TV Shows:", content_counts.get("TV Show", 0))
print("Total Content:", len(df))
```

```python
content_counts.plot(
    kind="bar",
    color=["skyblue", "orange"],
    figsize=(7, 5)
)

plt.title("Netflix Content Distribution")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
content_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    startangle=90
)

plt.title("Proportion of Movies and TV Shows")
plt.ylabel("")
plt.show()
```

```python
content_percent = (content_counts / len(df)) * 100

print("Content Proportions:")
print(content_percent.round(2))

print("\nMovies Percentage:",
      round(content_percent.get("Movie", 0), 2), "%")

print("TV Shows Percentage:",
      round(content_percent.get("TV Show", 0), 2), "%")
```

```python
movies = content_counts.get("Movie", 0)
tv_shows = content_counts.get("TV Show", 0)
total = len(df)

movie_pct = (movies / total * 100) if total else 0
tv_pct = (tv_shows / total * 100) if total else 0

print("KEY FINDINGS")
print("-" * 35)
print(f"Total Content: {total}")
print(f"Total Movies: {movies} ({movie_pct:.2f}%)")
print(f"Total TV Shows: {tv_shows} ({tv_pct:.2f}%)")

if movies > tv_shows:
    print("Finding: Movies are more numerous than TV Shows.")
elif tv_shows > movies:
    print("Finding: TV Shows are more numerous than Movies.")
else:
    print("Finding: Movies and TV Shows are equal in number.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset_Cleaned.csv")

print(df["country"].head())
```

```python
country_df = df.copy()

country_df["country"] = country_df["country"].fillna("Unknown")

country_df["country"] = country_df["country"].astype(str).str.strip()

country_df["country"] = country_df["country"].str.split(",")

country_df = country_df.explode("country")

country_df["country"] = country_df["country"].str.strip()

country_df = country_df[
    (country_df["country"] != "") &
    (country_df["country"].str.lower() != "unknown")
]

print(country_df[["title", "country"]].head(10))
```

```python
country_counts = country_df["country"].value_counts()

print(country_counts.head(10))
```

```python
top_10 = country_counts.head(10)

print("Top 10 Content-Producing Countries")
print(top_10)
```

```python
top_10.sort_values().plot(
    kind="barh",
    figsize=(10, 6),
    color="steelblue"
)

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
ranking = country_counts.head(10).reset_index()

ranking.columns = ["Country", "Content Count"]
ranking.index = ranking.index + 1
ranking.index.name = "Rank"

print(ranking)
```

```python
top_country = country_counts.index[0]
top_count = country_counts.iloc[0]

print("BUSINESS INSIGHTS")
print("-" * 35)

print(f"1. Top country: {top_country}")
print(f"2. Content count: {top_count}")

print(
    "3. The top 5 countries account for",
    round(country_counts.head(5).sum() / country_counts.sum() * 100, 2),
    "% of all country-title records."
)

print(
    "4. Country counts include titles associated with multiple countries."
)

print(
    "5. These results show catalog representation, not necessarily "
    "the audience size or popularity in each country."
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset_Cleaned.csv")

print(df[["title", "release_year"]].head())
```

```python
df["release_year"] = pd.to_numeric(
    df["release_year"], errors="coerce"
)

year_df = df.dropna(subset=["release_year"]).copy()

year_df["release_year"] = year_df["release_year"].astype(int)

print("Total valid records:", len(year_df))
print(year_df[["title", "release_year"]].head())
```

```python
yearly_releases = (
    year_df["release_year"]
    .value_counts()
    .sort_index()
)

print(yearly_releases.tail(15))
```

```python
top_years = yearly_releases.sort_values(
    ascending=False
).head(10)

print("Top 10 Release Years:")
print(top_years)
```

```python
yearly_analysis = yearly_releases.to_frame(
    name="Content_Count"
)

yearly_analysis["Change"] = (
    yearly_analysis["Content_Count"].diff()
)

yearly_analysis["Growth_Percent"] = (
    yearly_analysis["Content_Count"].pct_change() * 100
)

print(yearly_analysis.tail(15).round(2))
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    yearly_releases.index,
    yearly_releases.values,
    marker="o"
)

plt.title("Netflix Content Releases by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
top_years.sort_index().plot(
    kind="bar",
    figsize=(10, 5),
    color="teal"
)

plt.title("Top 10 Years by Content Releases")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
peak_year = yearly_releases.idxmax()
peak_count = yearly_releases.max()

latest_year = yearly_releases.index.max()
latest_count = yearly_releases.loc[latest_year]

print("BUSINESS INSIGHTS")
print("-" * 40)

print(f"Peak release year: {peak_year}")
print(f"Titles released in peak year: {peak_count}")

print(f"Latest release year in dataset: {latest_year}")
print(f"Titles for that year: {latest_count}")

print(
    "\nNote: Recent years may appear lower because "
    "the dataset may not include every newly released title."
)
```
